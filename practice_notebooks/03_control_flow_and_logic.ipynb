## 1. Making a Decision with if

An `if` statement runs a block of code only when a condition is `True`.

**Task:** Store an age in a variable, then print `"You can vote."` if the age is 18 or more.

````python
age = 31
if age >=18:
    print("You can vote.")
````

## 2. How Python Knows What's Inside a Block (Indentation)

Python uses indentation - the spaces at the start of a line - to show which lines belong inside an `if` block, instead of using brackets like some other languages. Every line inside the same block needs the same amount of indentation, usually 4 spaces.

**Task:** Store an age in a variable, then print two lines inside the same `if` block when the age is 18 or more.

````python
age = 17
if age >= 18:
    print("You can vote")
else:
    print("You can vote when you are 18 or older!")
````

## 3. Why Consistent Indentation Matters

If the lines inside a block aren't indented by the same amount, Python doesn't know they belong together and shows an error instead of running the code.

**Task:** Store an age in a variable, then write an `if` block where the second line uses one extra space of indentation compared to the first, and see what error Python shows.

````python
age = 22
if age >=18:
    print("you are old enough to vote.")
````

## 4. Choosing the Other Path with else

`else` runs its code when the `if` condition is `False`.

**Task:** Store an age in a variable, then print `"You can vote."` if the age is 18 or above, otherwise print `"You cannot vote."`

````python
age = 16
if age >=18:
    print('You can vote.')
else:
    print("You must be 18 or above before you can vote.")
````

## 5. Checking Several Conditions with elif

`elif` lets you check more conditions, one after another, before falling back to `else`.

**Task:** Store a score, then print `"A"` for 90 or above, `"B"` for 70 or above, and `"C"` for anything lower.

````python
score = 70
if score>=90:
    print("A")
elif score >=70:
    print("B")
else:
    print("C")
````

## 6. Repeating Code with a for Loop

A `for` loop repeats a block of code once for each character in a piece of text.

**Task:** Store a word, then use a `for` loop to print each of its letters on its own line.

````python
name = "Alexander"
for ch in name:
    print(ch)
````

## 7. Looping a Set Number of Times with range()

`range()` creates a sequence of numbers to loop over. With just one number, it counts starting from `0` up to (but not including) that number.

**Task:** Use `range(5)` to print the numbers `0` to `4`.

````python
for item in range(5):
    print(item)
````

## 8. Choosing Where range() Starts and Stops

Give `range()` two numbers to choose where the count starts and where it stops (still not including the last number).

**Task:** Use `range(1, 6)` to print the numbers `1` to `5`.

````python
for number in range(1,6):
    print(number)
````

## 9. Choosing a Step Size with range()

A third number tells `range()` how much to count up by each time, instead of just by `1`.

**Task:** Use `range(0, 11, 2)` to print every second number from `0` to `10`.

````python
for number in range(0,11,2):
    print(number)
````

## 10. Repeating Code While a Condition is True

A `while` loop keeps running its code for as long as a condition stays `True`.

**Task:** Store `0` in a variable, then use a `while` loop to print the numbers while it stays below `5`.

````python
number = 0
while number < 5:
    print(number)
    number +=1
````

## 11. Stopping a Loop Early with break

`break` stops a loop immediately, even if its condition is still `True`.

**Task:** Loop through the numbers 1 to 10, but stop as soon as you reach `6`.

````python
for i in range(1,11):
    print(i)
    if i == 6:
        break
````

## 12. Skipping a Step with continue

`continue` skips the rest of the current loop step and moves straight to the next one.

**Task:** Loop through the numbers 1 to 10 and print only the odd numbers, using `continue` to skip the even ones.

````python
for i in range(1,11):
    if i % 2 == 0:
        continue
    else:
        print(i)
````

## 13. Doing Nothing on Purpose with pass

`pass` is a placeholder that does nothing - handy when a line of code is required but there's nothing to do yet.

**Task:** Check if a number is negative; do nothing for negative numbers using `pass`, and print the number otherwise.

````python
number = -2
if number >=0:
    pass
else:
    print("negative number.")
````

## 14. Loops Inside Loops

A loop can contain another loop inside it - the inner loop finishes all its steps for each single step of the outer loop.

**Task:** Use two nested `for` loops to print a 3 by 4 grid of stars, with each row on its own line, so it looks like this:
```
* * * * 
* * * * 
* * * * 
```

````python


for num in range(1,4):
    line = ""
    for i in range(1,5):
        line += "* "
    print(line)
````

## 15. Putting It Together: Counting Characters

Let's combine loops, `continue`, and `break` in one practical example.

**Task:** Store the text `"I am Narender. I work for ABC company"` in a variable. First using a `for` loop:
1. Count every character in the text.
2. Do it again, but use `continue` to skip spaces so the count excludes them.
3. Extend that, and use `break` to stop counting as soon as you reach the first `.`.

Then repeat all three steps using a `while` loop instead.

````python
line = "I am narender. I work for ABC company"
count = 0
for ch in line:
    count += 1
print(count)
##########################
count = 0
for ch in line:
    if ch == " ":  ## omit spaces
        continue
    count += 1
print(count)
#######################################
count = 0
for ch in line:
    if ch == ".":  
        break

    if ch == " ":
        continue
    count += 1
print(count)
````

## Recap

- Python uses indentation (consistent spacing at the start of a line) to show which lines belong inside a block, instead of brackets - and every line in a block needs the same amount of indentation.
- `if`, `elif`, and `else` let a program choose between different paths.
- A `for` loop repeats code once for each character in a piece of text, or a set number of times using `range()`.
- `range()` can take one number (count from `0`), two numbers (start and stop), or three numbers (start, stop, and step size).
- A `while` loop repeats code for as long as a condition stays `True`.
- `break` stops a loop early, `continue` skips to the next step, and `pass` is a placeholder that does nothing.
- Loops can be nested inside each other.
